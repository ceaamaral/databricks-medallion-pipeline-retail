```python
%sql
-- Validando a tabela criada
SELECT * 
  FROM medallion_pipe_retail.bronze.sales LIMIT 30;


```

```python
%sql
-- Vendas duplicadas
SELECT sale_id
      ,COUNT(1) AS QTD 
  FROM medallion_pipe_retail.bronze.sales
 GROUP BY sale_id
 HAVING COUNT(SALE_ID) > 1;
```

```python
%sql
-- Vendas nulas
SELECT sale_id
      ,COUNT(1) AS QTD 
  FROM medallion_pipe_retail.bronze.sales
 WHERE sale_id IS NULL 
 GROUP BY sale_id;
```

```python
%sql
-- Vendas negativas
SELECT amount
      ,sale_id
      ,COUNT(1) AS QTD 
  FROM medallion_pipe_retail.bronze.sales
 WHERE amount < 0 
 GROUP BY amount
         ,sale_id;
```

```python
%sql
-- Tipos de status de venda
select distinct (shipping_status)
  from medallion_pipe_retail.bronze.sales;

```

```python
%sql
-- Vendas status inválido
SELECT shipping_status
      ,COUNT(1) AS QTD 
  FROM medallion_pipe_retail.bronze.sales
 WHERE shipping_status = 'Unknown'
 GROUP BY shipping_status; 
 
```
